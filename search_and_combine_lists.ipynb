# Marker Repo - search and combine lists

## Loading packages

```python
import markerrepo.marker_repo as mr
import markerrepo.wrappers as wrap
```

# Search database

## Return search results as metadata 

```python
results = mr.guided_search(out="metadata")
```

## Return search results as marker list DataFrame

```python
results = mr.guided_search(out="marker_list")
```

# Combine lists

Example 1: Combine all lists which contain the keywords.

```python
# Save search results in "df"
keywords = {"List name": "panglao", "Organism name": "mouse"}
df = mr.search_db(mr.get_db(), keywords, case_sensitive=False, exact=False, out="marker_list")
display(df)

# Export marker list
mr.export_marker_list(df)
```

Example 2: Use guided search instead + wrapper instead

```python
# Perform guided search to select marker lists
df = mr.guided_search(out="marker_list")

# Convert selected marker lists to panglao style list
wrap.convert_markers(df=df, style="panglao", path=".", file_name="guided_wrapper")
```

# Wrapper functions

Specify keywords

```python
keywords = {"List name": "panglao", "Organism name": "mouse"}
```

Use the keywords parameter to create a new marker list in "two column" style. This format can be used for custom cell type annotations via SCSA, for example.

```python
wrap.convert_markers(keywords=keywords, style="two_column", path=".", file_name="mouse_scsa")
```

Use the keywords parameter to create a new marker list in "panglao" style. This format can be used for the cell type annotations via the "custom script" of the single cell toolbox.

```python
wrap.convert_markers(keywords=keywords, style="panglao", path=".", file_name="mouse_panglao", organism="Mm")
```
