# Marker Repo - search and combine lists

## Loading packages

```python
import marker_repo as mr
import wrappers as wrap
```

## Setting

```python
# The path where the lists of the Marker Repo are stored - probable 'REPO_PATH/lists'
REPO_LISTS_PATH = "/mnt/workspace/mkessle/projects/annotate_by_marker_and_features/lists"
```

# Search database

```python
results = mr.guided_search(REPO_LISTS_PATH)
```

# Combine lists

Example 1: Use the results df created before.

```python
df = results

# Get UIDs and combine lists
uids = [int(idx) for idx in df.index]
combined_df = mr.combine_lists(REPO_LISTS_PATH, uids)

wrap.convert_markers(REPO_LISTS_PATH, df=combined_df, style="panglao", file_name="results_wrapper")
```

Example 2: Combine all lists which contain the keywords.

```python
# Save search results in "df"
keywords = {"List name": "panglao", "Organism name": "mouse"}
df = mr.search_db(mr.get_db(REPO_LISTS_PATH), keywords, case_sensitive=False, exact=False)

# Get UIDs and combine lists
uids = [int(idx) for idx in df.index]
combined_df = mr.combine_lists(REPO_LISTS_PATH, uids)
display(combined_df)

# Export marker list
mr.export_marker_list(combined_df)
```

Example 3: Use guided search instead + wrapper instead

```python
# Perform guided search to select marker lists
df = mr.guided_search(REPO_LISTS_PATH)

# Get UIDs and combine lists
uids = [int(idx) for idx in df.index]
combined_df = mr.combine_lists(REPO_LISTS_PATH, uids)

# Convert selected marker lists to score style list
wrap.convert_markers(REPO_LISTS_PATH, df=combined_df, style="panglao", path=".", file_name="guided_wrapper")
```
