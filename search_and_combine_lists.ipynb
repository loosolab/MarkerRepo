# Marker Repo - search and combine lists

## Loading packages

```python
import marker_repo as mr
```

## Setting

```python
# The path where the lists of the Marker Repo are stored - probable 'REPO_PATH/lists'
REPO_LISTS_PATH = "/mnt/workspace/mkessle/projects/annotate_by_marker_and_features/lists"
```

## Get and show database

```python
db = mr.get_db(REPO_LISTS_PATH)
display(db)
```

## Search database

Enter dictionary with key: column and value: keyword or enter a string to search the whole database:<br>
keywords = {"list_name": "Heart"} -> take only the column "list_name" into account<br>
keywords = {"submitter_name": "Kessler", "tags_gender": "mixed"}  -> take only the columns "submitter_name" and "gender" into account<br>
keywords = "WT" -> take all columns into account<br>

### Examples

"Heart" must be a substring of the column list_name.

```python
keywords = {"list_name": "Heart"}
mr.search_db(db, keywords, case_sensitive=False, exact=False)
```

"Kessler" must be a substring of submitter_name and "mixed" of tags_gender. Consider case sensitivity as well.

```python
keywords = {"submitter_name": "Kessler", "tags_gender": "mixed"} 
mr.search_db(db, keywords, case_sensitive=True, exact=False)
```

"WT" must appear in any column.

```python
keywords = "WT"
mr.search_db(db, keywords, case_sensitive=True, exact=True)
```

## Combine lists

Example 1: Combine all lists which contain the keyword "Genes".

```python
# Save search results in "df"
keywords = "Genes"
df = mr.search_db(db, keywords, case_sensitive=False, exact=True)

# Get UIDs and combine lists
uids = [int(idx) for idx in df.index]
combined_df = mr.combine_lists(REPO_LISTS_PATH, uids)
display(combined_df)

# Drop duplicates (redundant if you want to submit the new list using the submit_lists notebook)
markers_filtered = combined_df.drop_duplicates()
display(markers_filtered)

# Export marker list
mr.export_marker_list(REPO_LISTS_PATH, "all_genes", combined_df)
```
